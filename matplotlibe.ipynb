```python
import matplotlib.pyplot as plt
x = [0 , 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
plt.plot(x ,y, marker='o', label = ("data point"))

plt.title("plot graph")
plt.xlabel("x line")
plt.ylabel("y line")

plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
x = ['mon', 'sat', 'wed', 'thus']
y = [120, 160, 190, 206]
plt.plot(x ,y, marker='o', label = ("data point"))

plt.title("bar graph")
plt.xlabel("day")
plt.ylabel("total amount")

plt.legend()

plt.show()
```
